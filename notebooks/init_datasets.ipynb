Calculating `Statistic Parity Difference` and `Disparate Impact` in obesity data frame

Both are part of the fairness metrics suite

```python
def fairness_test(df, sensitive_attribute, target_col, positive_classes):
    y_bin = df[target_col].isin(positive_classes).astype(int)

    rates = y_bin.groupby(df[sensitive_attribute]).mean()

    p_female = rates.get("Female", float("nan"))
    p_male = rates.get("Male", float("nan"))

    spd = p_female - p_male # Statistic Parity Difference
    di = p_female - p_male if p_male > 0 else float("inf") # Disparate Impact
    
    return p_female, p_male, spd, di
```

```python
results = []
for name, positives in target_categories.items():
    p_f, p_m, spd, di = fairness_test(df, 
                                      sensitive_attribute="gender", 
                                      target_col="target", 
                                      positive_classes=positives
                                    )
    results.append({
        "task": name,
        "positives": ", ".join(positives),
        "P(Y=1|Female)": p_f,
        "P(Y=1|Male)": p_m,
        "SPD (F - M)": spd,
        "DI (F / M)": di,
    })

fairness_screen = pd.DataFrame(results)
fairness_screen
```

```python
df = pd.read_csv(raw_data_path / "arrhythmia/arrhythmia.data", header=None)
```

```python
df.shape
```

```python
target_raw = df.iloc[:,-1]
df["target"] = target_raw.apply(lambda x: 0 if x == 1 else 1)

# Drop original column
df.drop(columns=df.columns[-2], inplace=True) 

print(df["target"].value_counts())
```

```python
# Check for empty values in df columns

for col in df.columns:
    n_q = (df[col] == "?").sum()
    if n_q > 0:
        print(f"Found {n_q} values '?' in column {col}")
```

```python
# Due to the amount of empty values, col 13 will be dropped
# Other columns will gain input values
df = df.drop(columns=[13])

cols_to_fill = [10, 11, 12, 14]

# Swaps "?" for NaN
df[cols_to_fill] = df[cols_to_fill].replace("?", np.nan)

# Convert to Float
df[cols_to_fill] = df[cols_to_fill].astype(float)

# Input median
for col in cols_to_fill:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)

df = df.reset_index(drop=True)
```

```python
sex_col = 1

df[sex_col] = pd.to_numeric(df[sex_col], errors="coerce")

df[sex_col] = df[sex_col].map({0: 1, 1: 0})

df.rename(columns={sex_col: "sex"}, inplace=True)
print(df["sex"].value_counts())
```

```python
age_col = 0
df.rename(columns={age_col: "age"}, inplace=True)
print(df["age"].value_counts())
```

```python
df.to_csv(processed_data_path / f"11_arrhythmia_pp.csv")
```
